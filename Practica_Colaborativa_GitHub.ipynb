# Práctica colaborativa: NumPy + Pandas con Google Colab y GitHub

## Propósito
Esta actividad tiene como objetivo practicar ejercicios básicos de **NumPy** y **Pandas**, al mismo tiempo que se utiliza la integración entre **Google Colab y GitHub** para trabajar de forma colaborativa y secuencial sobre un mismo notebook.

## Modalidad de trabajo
- Trabajen en parejas.
- Ambos integrantes deben estar conectados por **Zoom**.
- Durante toda la actividad, la persona que esté programando debe **compartir pantalla**.
- El trabajo se realizará sobre **un único archivo `.ipynb` almacenado en GitHub**.
- **Estudiante A** resuelve primero los ejercicios 1 y 2.
- Al finalizar el ejercicio 2, Estudiante A guarda el notebook actualizado en GitHub.
- **Estudiante B** abre desde GitHub la versión actualizada y resuelve los ejercicios 3 y 4.
- Al finalizar, Estudiante B guarda nuevamente el notebook en GitHub.

> **Importante:** No trabajen simultáneamente sobre dos copias diferentes del notebook. El objetivo es practicar un flujo de trabajo secuencial utilizando GitHub como repositorio compartido.

## Flujo de trabajo con Google Colab y GitHub

1. Una persona del grupo debe tener el notebook dentro de un repositorio de GitHub.
2. Abra el archivo desde Google Colab usando **Archivo > Abrir cuaderno > GitHub**.
3. Busque el repositorio y abra este notebook.
4. Estudiante A resuelve los ejercicios 1 y 2.
5. Guarde los cambios usando **Archivo > Guardar una copia en GitHub**.
6. Estudiante B debe volver a abrir desde GitHub la versión más reciente del notebook.
7. Estudiante B resuelve los ejercicios 3 y 4.
8. Guarde nuevamente la versión final en GitHub.

### Recomendación para los commits

- Segundo guardado: `Ejercicios 3 y 4 completados - Nombre Estudiante B`
